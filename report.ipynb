### Introduction

*“The European airport community is a key enabler of economic and social growth, resilience, security, and Europe’s global competitiveness”* states the European Commission’s Directorate-General for Mobility and Transport. This suggests that disruptions within this European airport community can affect economic and social growth, resilience, security and Europe’s global competitiveness. One of such disruptions is flight delays. Flight delays are a remaining issue that can have substantial negative consequences, such as an increase in operational costs for airports, enhanced environmental impact (Pastorini and Zanin, 2022), passenger dissatisfaction, and a negative impact on the perceived reliability of European airports. To mitigate these effects, a thorough analysis of the causes and prediction of flight delays is essential. Such analyses can yield valuable insights that support the optimisation of airport operations and the enhancement of their overall resilience. 

Previous studies have already analysed the causes and effects of flight delays. For instance, Ashford et al. (2013) and Cohen and Paul (2003) showed how delays in operations at airports can lead to delayes of dependent operations. This, in turn, may result in delayed flight departures (Cohen and Paul, 2003). Also, studies have examined external factors like weather (Rodríguez-Sanz et al., 2022) and air traffic management (Makhtoumi, 2023). Quantitative studies have analysed delay patterns and prediction models (for instance, see: Güvercin et al., 2021; Shao et al., 2019). 

However, an integrated overview of the causes and predictions of delays at airports across Europe for recent, post-pandemic years has not been found. As a result, there is a gap in understanding of how airport-related factors, temporal and geographic variations, influence flight delay patterns in across Europe’s airports in recent years.  

### Research Objectives

This project aims to address this gap by offering a **quantitative analysis of flight delays at the twenty European airports with the most flights**, during the years **2023** and **2024**. The twenty busiest airports are selected based on the dataset ‘’AirportTraffic’’ provided by Eurocontrol. The busiest airports have been investigated as these might have a high impact on the overall air traffic, and delays at these airports might have a larger impact. The years 2023 and 2024 are chosen as these years are assumed not to be impacted by the COVID-19 pandemic. The focus is on airport-related delays; en-route delays are out of the scope of this research. This decision is made as many delays are caused by inefficiencies at airports.  

The research aims to answer the following main research question: 

> What are the main airport-related factors driving flight delays across the twenty busiest European airports in 2023–2024, and how accurately can these factors be used to predict the likelihood of potential flight delays through quantitative modelling and data visualisation? 

To answer this research question, sub-questions are formulated: 

1. What type of delays are occurring most at the airports? 

2. Which airports experience the highest, the lowest delay, and most consistent delay rates? 

3. How do delays relate to traffic volume? 

4. How do delays distribute geographically across Europe, and which airports stand out as persistent “delay hotspots”? 

5. What model can be used to predict whether an airport is prone to delays? 

All datasets used in this research were obtained from Eurocontrol, a European organisation dedicated to supporting and coordinating aviation across Europe. As part of their commitment to transparency and research, Eurocontrol provides a wide range of publicly accessible data. Certain datasets, however, are restricted and available only upon request. One such dataset is the “IATA” dataset, which serves as a primary source for this study. The remaining datasets utilised are publicly available.

The scatter plot reveals a positive relationship between flight volume and total delay, confirming that airports handling more flights generally accumulate higher overall delays. - **Congested high traffic airports**:

They show particularly high cumulative delays, reflecting congestion and operational complexity at large intercontinental hubs.  

- **Efficient high-volume performers**: 
Madrid (LEMD) and London Heathrow (EGLL) manage very high traffic volumes but maintain comparatively lower total delays than peers of similar size 

- **Medium underperformers**: 
Munich (EDDM), Rome (LIRF) and Lisbon (LPPT) have higher delay total (around 1 to 1.5 million minutes) despite its lower to moderate traffic volumes. This indicates potential capacity limitations or operational inefficiencies. 

- **Smaller but resilient airports**: 
Airports such as Manchester (EGCC), Vienna (LOWW), and Brussels (EBBR) operate fewer than 250 000 flights yet maintain total delays below 0.6 million minutes, reflecting operational efficiency at moderate capacity.

## Conclusions

This study set out to identify the main airport-related factors driving flight delays across the twenty busiest European airports in 2023–2024 and to evaluate how accurately these factors can predict delay likelihood through quantitative analysis and modelling. We can address the main research question:
“What are the main airport-related factors driving flight delays across the twenty busiest European airports in 2023–2024, and how accurately can these factors be used to predict the likelihood of potential flight delays through quantitative modelling and data visualisation?”

The analyses conducted throughout this study provide valuable insights into the causes and patterns of flight delays at Europe’s busiest airports. The airport flight delays are driven primarily by airline, airport, and en-route factors, while weather and government-related causes contribute to a smaller proportion of total delay minutes. However, within air traffic flow management data, weather emerges as the dominant cause, highlighting its growing impact under certain conditions. When looking into delay occurrences, a clear seasonal pattern is evident, with delay ratios peaking during the summer months when traffic volumes are highest. Delay ratios increase from approximately 3.5% in winter to 5.5% in summer. 

In terms of punctuality performance, arrival times are more reliable than departure times, suggesting that airlines often manage to recover delays during flight operations. The delay intensity and consistency analysis revealed that Oslo - Gardermoen had the best performance, consistently maintaining the lowest average monthly delay rate (below 10 min/flight), while Lisbon faced the highest average rates. In terms of predictability, Frankfurt demonstrated the most stable performance, whereas Antalya showed the most volatility, indicating the greatest inconsistency in its monthly delay rates.

A positive relationship was observed between flight volume and total delay minutes, confirming that high traffic typically lead to greater cumulative delays. However, this relationship was not perfectly linear. Major hubs such as Istanbul, Amsterdam, Paris Charles de Gaulle, and Frankfurt face extensive congestion, while London Heathrow and Madrid-Barajas manage similar volumes more efficiently.

Geospatial visualizations highlighted clear seasonal and regional variations. During the summer months, airports across southern Europe showed significant increases in both flight activity and total delay time and average delay per movement. This pattern aligns with the tourism peak happening in the summer holiday season.

Two predictive models — logistic regression and random forest regression — were developed to forecast whether an airport would be “delay-prone.” Both models achieved strong predictive performance, with accuracies of 78% and 81%, respectively. Logistic regression was identified as the more interpretable and reliable model, particularly given the dataset size and its linear relationships.


## Discussion

Overall, the findings for the research question are satisfactory and provide meaningful insights into the topic. However, analysing delays in aviation involves a wide range of interrelated factors, making it a complex issue that would require years of comprehensive study. For the purpose of this research, the scope was therefore narrowed to a selection of datasets made publicly available by Eurocontrol. Nonetheless, public data often contain significant gaps. For instance, the ATFM_delay dataset reports daily delays per airport, but many entries are missing, which can lead to incomplete or biased results. This limitation became apparent during the analysis, prompting a request for access to an additional, non-public dataset, which proved to be much more complete.

However, a considerable portion of the analysis code had already been developed based on the earlier, incomplete datasets. Adapting it to the new dataset proved challenging and, in some cases, not entirely feasible, as the new dataset, while more complete, contained less detailed information overall compared to the original ones.

Another challenge encountered during the data collection process was the inconsistency between datasets. In some cases, the total number of daily flights differed across datasets, even though all were provided by Eurocontrol. Consequently, combining multiple datasets may not always produce a fully accurate or realistic representation, as discrepancies between reported figures can affect the validity of the results.

For future research on aviation delays, it would be valuable to examine a broader range of variables, including weather patterns, air traffic management efficiency, and airport infrastructure capacity. Combining these with more complete, and more detailed data could provide a more comprehensive understanding of the underlying causes.
